##  Creating Strings

```python
s = 'hello'
s = "hello"
# multiline strings
s = '''hello'''
s = """hello"""
print(s)
```

## Accessing Substring from a String

```python
# Positive Indexing
s = 'hello world'
print(s[0])
```

```python
# Negative Indexing
s = 'hello world'
print(s[-1])
```

```python
# Slicing 
s = 'hello world'
print(s[0:6:2])
print(s[6:0:-2])
```

```python
print(s[::-1])
```

```python
s = 'hello world'
print(s[-1:-6:-1])
```

## Editing and Deleting in Strings

```python
s = 'hello world'
s[0] = 'H'
# Python strings are immutable ( which means no editing, changes are not allowed in python)
```

```python
s = 'hello world'
del s
print(s)
```

```python
s = 'hello world'
del s[-1:-5:2]
print(s)
```
