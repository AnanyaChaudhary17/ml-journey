### Creating a list 

```python
# Empty
print([])
# 1D --> homogeneous list
print([1,2,3,4,5])
# 2D
print([1,2,3,[4,5]])
# 3D
print([[[1,2],[3,4]],[[5,6],[7,8]]])
# Hetrogeneous
print([1,True,5.6,5+6j,'Hello'])
# Using Type conversion
print(list('hello'))

```

### Accessing items from a list

```python
# Indexing
L = [1,2,3,[4,5]]
print(L[0]) #positive indexing L-->R
print(L[3][0])
print(L[-3]) #negative indexing R-->L
print(L[-1][-2])

L = [[[1,2],[3,4]],[[5,6],[7,8]]]
print(L[1][1][0])
print(L[0][0][1])

# Slicing
L = [1,2,3,4,5,6]

print(L[-3:])
print(L[0::2])
print(L[::-1]) #list reverse

# 2D, 3D, ya jitne nD list h jitne dimenions h utne [] lga kr kaam krenge 

```

### Adding Items to a list

```python
# append
L = [1,2,3,4,5]
L.append(True)
print(L)
# append add single item in last
```

```python
# extend
L = [1,2,3,4,5]
L.extend([6,7,8])
print(L)
#extend multiple item ko ek sath list me add krne k liye use hota h last me
```

```python
# insert
# desired place pr add krta h insert ek br me ek chiz hi add hogi multiple chiz add nhi hogi

L = [1,3,4,5,6]
L.insert(1,100)
print(L)

```

### Editing items in a list

```python
L = [1,2,3,4,5]

# editing with indexing
L[-1] = 500

# editing with slicing
L[1:4] = [200,300,400]

print(L)
```

### Deleting items from a list

```python
# del
L = [1,2,3,4,5]
#indexing
del L[-1]
print(L)

# slicing
del L[1:3]
print(L)
```

```python
# remove
# jha value se dlt krte h
L = [1,2,3,4,5]
L.remove(5)
print(L)
```

```python
# pop
# index pkd kr dlt krta h if index provide nhi krengee to last item dlt krta h.

L = [1,2,3,4,5]

L.pop()
print(L)
```

```python
# clear 
L = [1,2,3,4,5]
L.clear()
print(L)
```

### Operations on Lists

- Arithmetic
- Membership
- Loop

```python
# Arithmetic (+ , *)
L1 = [1,2,3,4]
L2 = [5,6,7,8]

# Concatenation/Merge
print(L1 + L2)

# multiplication
print(L1*3)
```

```python
# Membership

L1 = [1,2,3,4,5]
L2 = [1,2,3,4,[5,6]]

print(5 not in L1)
print([5,6] in L2)
```

```python
# Loops
L1 = [1,2,3,4,5]
L2 = [1,2,3,4,[5,6]]
L3 = [[[1,2],[3,4]],[[5,6],[7,8]]]

for i in L3:
    print(i)

```

### List Functions

```python
# len/min/max/sorted
L = [2,1,5,7,0]

print(len(L))
print(min(L))
print(max(L))
print(sorted(L))
```

```python
# count

L = [1,2,1,4,5,3,7,4]
L.count(5)
```

```python
# index
L = [1,2,1,4,5,3,7,4]
L.index(3)
```

```python
# reverse

L = [1,2,1,4,5,3,7,4]
# permanently reverse the list
L.reverse()
print(L)
```

```python
# sort (vs sorted)
L = [2,1,5,7,0]
print(L)
print(sorted(L)) # sorted is temporarily sort
print(L)
L.sort() # permanently sort
print(L)
```

```python
# copy --> shallow copy 
L = [2,1,5,7,0]
print(L)
print(id(L))
L1 = L.copy()
print(L1)
print(id(L1))
```

### List comprehension

```python
# Add 1 to 10 numbers to a list
L = []

for i in range(1,11):
    L.append(i)
print(L) 

# List comprehension
L = [i for i in range(1,11)]
```

```python
# scalar multiplication on a vector

v = [2,3,4]
s = -3
# [-6,-9,-12]

[s*i for i in v]
```

```python
# Add squares
L = [1,2,3,4,5]

[i**2 for i in L]
```

```python
# Print all numbers divisible by 5 in the range of 1 to 50

[i for i in range(1,51) if i%5 == 0]
```

```python
# find languages which start with letter p
languages = ['java', 'python', 'php', 'c', 'javascript']

[language for language in languages if language.startswith('p')]
```

```python
# Nested if with List Comprehension
basket = ['apple', 'guava','cherry','banana']
my_fruits =  ['apple', 'kiwi', 'grapes', 'banana']

# aadd new list from my_fruits and items if the fruit exists in basket and also starts with 'a'

[fruit for fruit in my_fruits if fruit in basket if fruit.startswith('a')]
```

```python
# cartesian products -> list comprehension on 2 lists together 
L1 = [1,2,3,4]
L2 = [5,6,7,8]

[i*j for i in L1 for j in L2]
```

```python
# Print a (3,3) matrix using list comprehension -> Nested List comprehension
[[i*j for i in range(1,4)] for j in range(1,4)]
```

### 2 ways to traverse a list

- itemwise
- indexwise 

```python
# itemwise 
L = [1,2,3,4]

for i in L:
    print(i)
```

```python
# indexing
L = [1,2,3,4]

for i in range(0,len(L)):
    print(i)
```
