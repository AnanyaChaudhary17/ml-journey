# Operators in Python


```python
# Arithmetic Operators
print(5+5)

print(5-6)

print(5/2)

print(5//2) #Interger division

print(5%2)

print(5**2) #power off operator
```

```python
# Relational Operators
print(4>5)

print(4<5)

print(4<=4)

print(4==4)

print(4!=4)
```

```python
# Logical Operators
print(1 and 0)

print(0 or 1)

print(not 1)
```

```python
# Bitwise Operator

#bitwise and
print(2 & 3)

#bitwise or 
print(2 | 3)

#bitwise xor
print(2 ^ 3)

#bitwise not
print(~1)

#bitwise left operator
print(4 >> 2)

#bitwise right operator
print(5 << 2)


```

```python
# Assignment Operator
# =
# a = 2
a = 2
a %= 2 #a =a % 2
print(a)

```

```python
# Membership Operators
# in/not in
print('D' not in 'Delhi')
print(1 in [2,3,4,5,6])
```

# if-else in Python

```python
# login program and indentation
# email -> ananya.code@gmail.com
# password -> 1234
email = input('enter email')
password =  input('enter password')

# if condition:
    #code
    # else:
    #code
if email == 'ananya.code@gmail.com' and password == '1234':
   print('Welcome')
elif email == 'ananya.code@gmail.com' and password != '1234':
    #tell the user
    print('Incorrect password')
    password = input('enter password again')
    if password  ==  1234:
        print('Welcome, finally')
    else:
        print('beta tumse na ho paega')
else:
    print('Not correct')
```

# Loops in Python

```python
# While Loop
# While Loop example -> program to print the table

number = int(input('enter the number'))
i =  1

while i<11:
  print(number, '*',i,'=',number * i)
  i += 1




```

```python
# while loop with else

x = 1
while x < 3:
    print(x)
    x +=1
else:
    print('limit crossed')
```

```python
# guessing game
# generate a random integer between 1 and 100

import random
jackpot = random.randint(1,100)

guess = int(input('guess kro'))
counter = 1
while guess != jackpot:
 if guess < jackpot:
    print('galat!guess higher')
 else:
    print('galat!guess lower')

 guess = int(input('guess kro'))
 counter += 1
else:
    print('correct guess')
    print('attempts',counter) 
```

# Sequence sum
1/1! + 2/2! + 3/3!...

```python
n = int(input('enter n'))

result = 0
fact = 1

for i in range(1,n+1):
    fact = fact * i
    result = result + i/fact
print(result)   
```

### Pattern 1
### *
### **
### ***

```python
rows = int(input('enter number of rows'))

for i in range(1,rows+1):
    for j in range(1,i+1):
        print('*',end='')
    print()   
```

### Pattern 2
1  
121  
12321  
1234321  

```python
rows = int(input('enter number of rows'))

for i in range(1,rows+1):
    for j in range(1,i+1):
        print(j,end='')
    for k in range(i-1,0,-1):
        print(k,end='')
    print()   
```

### Loop Control Statement 
- Break
- Continue
- Pass


```python
for i in range(1,10):
    if i == 5:
        break
    print(i)    
```

```python
lower = int(input('enter lower range'))
upper = int(input('enter upper range'))

for i in range(lower,upper+1):
    for j in range(2,i):
        if i%j == 0:
            break
    else:
        print(i)
```

```python
# continue
for i in range(1,10):
    if i == 5:
        continue
    print(i)
```

```python
for i in range(1,10):
    pass
```
